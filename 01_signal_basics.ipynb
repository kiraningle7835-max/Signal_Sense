## 1. Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal  # used for square, sawtooth and triangle waves

plt.rcParams['figure.figsize'] = (10, 3)
```

## 2. Time axis and sampling

A computer stores a signal as samples. With sampling rate `fs = 1000` Hz we take 1000 points every second.

```python
fs = 1000                          # samples per second (Hz)
t = np.arange(0, 1, 1/fs)          # 1 second of time stamps
print(len(t), 'samples; first 5 time stamps:', t[:5])
```

## 3. A sine wave

`x(t) = A * sin(2*pi*f*t + phase)`. Try 50 Hz, the frequency of mains supply in India.

```python
A, f, phase = 1.0, 5, 0           # 5 Hz so we can see cycles clearly
x = A * np.sin(2*np.pi*f*t + phase)

plt.plot(t, x)
plt.title('Sine wave: 5 Hz'); plt.xlabel('Time (s)'); plt.ylabel('Amplitude'); plt.grid(alpha=.3)
plt.show()
```

## 4. Four standard waveforms

```python
sine     = np.sin(2*np.pi*f*t)
square   = signal.square(2*np.pi*f*t)
sawtooth = signal.sawtooth(2*np.pi*f*t)
triangle = signal.sawtooth(2*np.pi*f*t, width=0.5)

fig, axes = plt.subplots(2, 2, figsize=(10, 6), sharex=True)
for ax, y, name in zip(axes.ravel(), [sine, square, sawtooth, triangle],
                       ['Sine', 'Square', 'Sawtooth', 'Triangle']):
    ax.plot(t, y); ax.set_title(name); ax.grid(alpha=.3)
plt.tight_layout(); plt.show()
```

## 5. Adding noise

Real signals are never clean. We add **Gaussian noise** with a chosen standard deviation.

```python
rng = np.random.default_rng(0)     # fixed seed so results are repeatable
noise = 0.5 * rng.standard_normal(t.size)
noisy_sine = sine + noise

plt.plot(t, noisy_sine, alpha=.6, label='Noisy')
plt.plot(t, sine, 'k', lw=2, label='Clean')
plt.legend(); plt.title('Sine wave with Gaussian noise'); plt.show()
```

## 6. Measuring noise: SNR

`SNR (dB) = 10 * log10(signal power / noise power)`, where power = mean of the squared signal. A higher SNR means a cleaner signal.

```python
def snr_db(clean, noise):
    return 10 * np.log10(np.mean(clean**2) / np.mean(noise**2))

for sigma in [0.1, 0.5, 1.0]:
    n = sigma * rng.standard_normal(t.size)
    print(f'noise std = {sigma}: SNR = {snr_db(sine, n):.1f} dB')
```

## 7. Aliasing demo

What if we sample too slowly? A 90 Hz sine sampled at only 100 Hz looks like a different, slower wave.

```python
f_true, fs_low = 90, 100
t_fine = np.arange(0, 0.2, 1/10000)
t_low  = np.arange(0, 0.2, 1/fs_low)

plt.plot(t_fine, np.sin(2*np.pi*f_true*t_fine), alpha=.5, label='True 90 Hz')
plt.plot(t_low, np.sin(2*np.pi*f_true*t_low), 'ro-', label='Sampled at 100 Hz')
plt.legend(); plt.title('Aliasing: the samples trace a false, slower wave'); plt.show()
```
